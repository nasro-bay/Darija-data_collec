# Algerian Darija Arabizi Transliteration Demo

This notebook demonstrates the stochastic rule-based transliterator that converts Algerian Darija from Arabic script to Arabizi (Latin script). It uses the character mappings, context-sensitive Madd rules, and the word lexicon derived from YouTube comment data.

```python
import json
import random
from transliterate import transliterate

# Set random seed for reproducibility
random.seed(42)
```

## 1. Quick Verification / Basic Examples

```python
examples = [
    "في",
    "على",
    "عليهم",
    "ربي يحفظك خويا",
    "راني عارف واش كاين في الدار"
]

for ex in examples:
    print(f"Arabic: {ex}")
    print(f"Arabizi: {transliterate(ex)}")
    print("-" * 30)
```

## 2. Load and Transliterate sample comments from `arabic_comments.jsonl`

```python
arabic_comments_file = "arabic_comments.jsonl"

comments = []
with open(arabic_comments_file, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            comments.append(json.loads(line))

print(f"Loaded {len(comments)} Arabic comments.")
```

## 3. Translate and Compare
Let's select 15 comments and print them side-by-side with their stochastic Arabizi translations.

```python
sampled_comments = random.sample(comments, min(15, len(comments)))

for i, comment in enumerate(sampled_comments, 1):
    original_text = comment["text"]
    arabizi_text = transliterate(original_text)
    print(f"{i}. Original Arabic:")
    print(f"   {original_text}")
    print(f"   Arabizi Transliteration:")
    print(f"   {arabizi_text}")
    print("=" * 50)
```
